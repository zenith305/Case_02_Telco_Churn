## Разведочный анализ (EDA)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/cleaned_data.csv')

# 1. Проверим дисбаланс классов (сколько ушло, сколько осталось)
print(df['Churn'].value_counts(normalize=True))

# 2. Посмотрим, как стаж (tenure) влияет на уход
sns.histplot(data=df, x='tenure', hue='Churn', multiple='stack')
plt.show()
```

## Boxplot

```python
# Настраиваем размер графиков, чтобы было видно
plt.figure(figsize=(15, 5))

# График 1: Влияние типа контракта
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Contract', hue='Churn')
plt.title('Уход в зависимости от типа контракта')

# График 2: Влияние ежемесячной платы
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='Churn', y='MonthlyCharges')
plt.title('Ежемесячная плата и уход')

plt.show()
```
